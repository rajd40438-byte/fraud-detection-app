```python
"""
Generates a small sample CSV to test the Streamlit app's "Upload CSV" tab.
Run this AFTER eda_and_modeling.py has run at least once (needs data/creditcard.csv).
"""

import pandas as pd

df = pd.read_csv(r"C:\Users\rkdas\OneDrive\Documents\Desktop\credit_card_fraud_detection\creditcard.csv\creditcard.csv")

# Take 20 legit + all available fraud rows from a slice, for a realistic mix
sample_legit = df[df["Class"] == 0].sample(20, random_state=42)
sample_fraud = df[df["Class"] == 1].sample(5, random_state=42)

sample = pd.concat([sample_legit, sample_fraud]).sample(frac=1, random_state=1)  # shuffle
sample = sample.drop(columns=["Class"])  # app expects no Class column, it predicts this

sample.to_csv("outputs/sample_for_upload.csv", index=False)
print("Saved outputs/sample_for_upload.csv —", len(sample), "rows")
```
